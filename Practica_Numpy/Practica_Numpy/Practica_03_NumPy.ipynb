# Práctica 3. Operaciones vectorizadas, máscaras booleanas y filtrado

Se analizarán mediciones simuladas de sensores ambientales. Se utilizarán operaciones vectorizadas y máscaras booleanas para filtrar observaciones sin recurrir a ciclos explícitos.

### 1. Importar NumPy

NumPy permite realizar operaciones numéricas sobre arreglos completos. Importa la biblioteca con el alias convencional `np`.

```python
import numpy as np
```

### 2. Crear los datos

El arreglo representa concentraciones de partículas PM2.5 registradas durante ocho mediciones. Los valores son ilustrativos y se expresan en µg/m³.

```python
pm25 = np.array([12.5, 18.2, 35.7, 22.1, 41.3, 16.8, 28.4, 52.6])
print(pm25)
```

### 3. Aplicar una operación vectorizada

Una operación vectorizada se aplica a todos los elementos sin escribir un ciclo `for`. Aquí se multiplica cada medición por 1.1.

```python
pm25_ajustado = pm25 * 1.1
print(pm25_ajustado)
```

### 4. Construir una máscara booleana

Una comparación sobre un arreglo genera valores `True` y `False`. La máscara identifica mediciones superiores a 25 µg/m³.

```python
mascara = pm25 > 25
print(mascara)
```

### 5. Filtrar observaciones

Al usar una máscara booleana dentro de los corchetes, NumPy devuelve únicamente los valores que cumplen la condición.

```python
mediciones_altas = pm25[mascara]
print(mediciones_altas)
```

### 6. Combinar condiciones

Las condiciones se combinan con `&` (AND) o `|` (OR). Cada condición debe escribirse entre paréntesis.

```python
rango_intermedio = pm25[(pm25 >= 15) & (pm25 <= 35)]
print(rango_intermedio)
```

### 7. Contar observaciones

`np.sum` suma valores booleanos como 1 y 0. Así se cuenta cuántas mediciones superan un umbral.

```python
cantidad_altas = np.sum(pm25 > 25)
print('Mediciones superiores a 25:', cantidad_altas)
```

### 8. Clasificar mediante una condición

`np.where(condición, valor_si_verdadero, valor_si_falso)` genera un arreglo con una etiqueta para cada medición.

```python
clasificacion = np.where(pm25 > 25, 'Alta', 'No alta')
print(clasificacion)
```

### 9. Obtener índices

`np.where` también puede devolver los índices donde una condición es verdadera.

```python
indices_altos = np.where(pm25 > 25)
print(indices_altos[0])
```

### 10. Analizar el subconjunto

Después de filtrar los datos, se pueden aplicar funciones estadísticas al subconjunto. Calcula el promedio de las mediciones altas.

```python
print('Promedio de mediciones altas:', np.mean(mediciones_altas))
```
