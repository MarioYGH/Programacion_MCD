# Práctica 5. Análisis integrado de datos numéricos con NumPy

Esta práctica integra creación, validación, transformación, filtrado y análisis estadístico de datos. El escenario simula el consumo eléctrico diario en cuatro edificios durante una semana.

### 1. Importar NumPy

Importa NumPy para realizar el análisis numérico.

```python
import numpy as np
```

### 2. Crear la matriz de consumo

Cada fila representa un edificio y cada columna un día de la semana. Los consumos, expresados en kWh, son datos simulados para fines didácticos.

```python
consumo = np.array([[120, 135, 128, 142, 150, 160, 145], [98, 105, 102, 110, 115, 118, 108], [155, 160, 158, 170, 175, 190, 180], [80, 85, 82, 88, 90, 95, 92]])
print(consumo)
```

### 3. Validar dimensiones y tamaño

Verifica las dimensiones y el número de observaciones antes de analizar la matriz.

```python
print('Dimensiones:', consumo.shape)
print('Número de datos:', consumo.size)
print('Número de edificios:', consumo.shape[0])
print('Número de días:', consumo.shape[1])
```

### 4. Calcular total y promedio por edificio

Con `axis=1`, se resume cada fila. El total semanal y el promedio diario permiten comparar el consumo de los edificios.

```python
total_edificio = np.sum(consumo, axis=1)
promedio_edificio = np.mean(consumo, axis=1)
print('Total semanal por edificio:', total_edificio)
print('Promedio diario por edificio:', promedio_edificio)
```

### 5. Analizar consumo por día

Con `axis=0`, se resume cada columna para identificar los días con mayor consumo conjunto.

```python
total_dia = np.sum(consumo, axis=0)
promedio_dia = np.mean(consumo, axis=0)
print('Total por día:', total_dia)
print('Promedio por día:', promedio_dia)
```

### 6. Identificar el edificio con mayor consumo semanal

`np.argmax` devuelve el índice del valor máximo. Los índices comienzan en cero.

```python
indice_mayor = np.argmax(total_edificio)
print('Índice del edificio con mayor consumo:', indice_mayor)
print('Consumo semanal máximo:', total_edificio[indice_mayor])
```

### 7. Filtrar consumos por encima de un umbral

Una máscara booleana extrae los registros que superan el umbral de 150 kWh.

```python
consumos_altos = consumo[consumo > 150]
print(consumos_altos)
print('Cantidad de registros altos:', consumos_altos.size)
```

### 8. Contar días de consumo alto por edificio

La matriz booleana `consumo > 150` puede sumarse por fila para contar cuántos días cada edificio superó el umbral.

```python
dias_altos_edificio = np.sum(consumo > 150, axis=1)
print(dias_altos_edificio)
```

### 9. Calcular variabilidad por edificio

La desviación estándar resume cuánto fluctúan los consumos diarios respecto a su promedio.

```python
desviacion_edificio = np.std(consumo, axis=1)
print(desviacion_edificio)
```

### 10. Simular una reducción del consumo

Supón que una medida de eficiencia reduce cada consumo en un 8 %. Una operación vectorizada calcula todos los valores ajustados.

```python
consumo_eficiente = consumo * 0.92
ahorro = consumo - consumo_eficiente
print('Consumo ajustado:')
print(consumo_eficiente)
print('Ahorro total estimado:', np.sum(ahorro))
```

### 11. Comparar consumo antes y después

Compara el consumo total original con el total después de aplicar la reducción para cuantificar el ahorro estimado.

```python
total_original = np.sum(consumo)
total_eficiente = np.sum(consumo_eficiente)
print('Total original:', total_original)
print('Total ajustado:', total_eficiente)
print('Ahorro:', total_original - total_eficiente)
```

### 12. Centrar los datos por edificio

Restar el promedio de cada fila permite estudiar las desviaciones respecto al comportamiento medio de cada edificio.

```python
promedios_fila = np.mean(consumo, axis=1, keepdims=True)
consumo_centrado = consumo - promedios_fila
print(consumo_centrado)
```
