Práctica Python
Creación y análisis de arreglos numéricos.
Objetivo: Aprender a crear arreglos con NumPy, identificar sus propiedades y realizar operaciones matemáticas básicas

1. Importar la libreria NumPy
import numpy: importa la biblioteca NumPy para utilizar sus funcionalidades.
as np: establece np como alias para escribir instrucciones más cortas.

```python

import numpy as np
```

Problema1.
Una empresa registra unidades de un producto que vendió durante 5 días consecutivos. Los datos son los siguientes:
Día/Unidades vendidas
Lunes/12
Martes/15
Miércoles/9
Jueves/18
Viernes/16
Objetivo: almacenar las ventas en un arreglo y analizar su comportamiento.

np.array(...): crea un arreglo de NumPy a partir de los datos proporcionados.
[12, 15, 9, 18, 16]: es una lista de Python que contiene las cinco ventas.
ventas = ...: asigna el arreglo a una variable llamada ventas.
print(ventas): muestra los valores almacenados
print(type(ventas)) indica que es un arreglo de n dimensiones.

```python
ventas = np.array([12, 15, 9, 18, 16])

print(ventas)
print(type(ventas))
```

Problema2.
Identificar las propiedades del arreglo.
size: indica cuántos valores contiene el arreglo
ndim: indica cuántos ejes tiene el arreglo
shape: devuelve una tupla con el tamaño de cada dimensión
dtype: indica qué tipo de valores almacena el arreglo.

```python

print("Número de elementos:", ventas.size)
print("Dimensiones del arreglo:", ventas.ndim)
print("Forma del arreglo:", ventas.shape)
print("Tipo de datos:", ventas.dtype)
```

Problema3. Realizar operaciones matemática para saber el comportamiento de las ventas durante 5 días. 
1. Total de ventas: sum()
2. Promedio mean()
3. Valores mínimo y máximo: max(), min()

```python

total = np.sum(ventas)

print("Total de unidades vendidas:", total)

promedio = np.mean(ventas)

print("Promedio diario de ventas:", promedio)

maximo = np.max(ventas)
minimo = np.min(ventas)

print("Venta máxima:", maximo)
print("Venta mínima:", minimo)
```

Problema4.
La empresa quiere analizar qué ocurriría si las ventas de cada día aumentaran en 3 unidades.

```python

ventas_nuevas = ventas + 3

print("Ventas originales:", ventas)
print("Ventas con aumento:", ventas_nuevas)
```

Problema5.
La empresa quiere estimar los ingresos diarios si cada unidad tiene un costo de venta de $25 a la venta.

```python

precio = 25
ingresos = ventas * precio

print("Ingresos por día:", ingresos)
```

Problema6.
La empresa quiere consultar una venta de un día en específico.
Los arreglos en Python inician con cero, por lo tanto, los índices para el arreglo ventas con 0,1,2,3,4, considerando que el número de elementos del arreglo es 5.

```python

print("Venta del lunes:", ventas[0])
print("Venta del miércoles:", ventas[2])
print("Venta del viernes:", ventas[4])
#el índice -1 hace referencia al último elemento del arreglo
print("Última venta:", ventas[-1])

```

Problema 7.
La empresa revisa sus registros e identifica que la venta del miercoles es incorrecta, debe ser 11 y no 9.


```python

print("Ventas antes de la corrección:", ventas)

ventas[2] = 11

print("Ventas después de la corrección:", ventas)
```

Problema 8. 
La empresa desea analizar una parte de sus ventas.
1. Las ventas de los primeros 3 días
2. Las ventas de jueves y viernes
3. Las ventas de miércoles a viernes

```python

primeros_tres_dias = ventas[0:3]

print(primeros_tres_dias)
```

```python

ultimos_dos_dias = ventas[3:5]

print(ultimos_dos_dias)
```

```python

desde_miercoles = ventas[2:]

print(desde_miercoles)
```

```python
ventas[1:4]
```

Explicación: Python los rangos se consideran de la siguiente manera [a,b), es decir, cerrado al inicio (se toma mayor o igual) y abierto al final (menor a)

Problema. 9
Realizar un análisis descriptivo de las ventas.
La empresa necesita un resumen que indique:
1. total de unidades vendidas
2. promedio diario
3. venta máxima y venta mínima
4. diferencia entre las ventas mínima y máxima.

```python

total = np.sum(ventas)
promedio = np.mean(ventas)
maximo = np.max(ventas)
minimo = np.min(ventas)
rango = maximo - minimo

print("RESUMEN DE VENTAS")
print("-----------------")
print("Total vendido:", total)
print("Promedio diario:", promedio)
print("Venta máxima:", maximo)
print("Venta mínima:", minimo)
print("Diferencia entre máximo y mínimo:", rango)
```
