# Práctica 2. Arreglos bidimensionales con NumPy

## Objetivo
Crear e interpretar arreglos bidimensionales con NumPy, consultar sus dimensiones y seleccionar valores mediante índices de filas y columnas.

## Contexto
Una tienda registra las unidades vendidas de tres productos durante cuatro días. Cada fila representará un producto y cada columna un día.

| Producto | Lunes | Martes | Miércoles | Jueves |
|---|---:|---:|---:|---:|
| Producto A | 12 | 15 | 11 | 18 |
| Producto B | 8 | 10 | 14 | 9 |
| Producto C | 20 | 17 | 16 | 22 |

## Paso 1. Importar NumPy

NumPy es una biblioteca de Python que permite trabajar eficientemente con arreglos numéricos. Para utilizar sus funciones, primero se debe importar la biblioteca. La convención habitual es asignarle el alias `np`, lo que permite escribir instrucciones más breves.

```python
import numpy as np
```

## Paso 2. Crear un arreglo bidimensional

Un arreglo bidimensional organiza los datos en filas y columnas, de manera similar a una tabla. Utilizaremos `np.array()` para crear una matriz con las ventas de tres productos durante cuatro días. En este ejemplo, cada fila corresponde a un producto y cada columna representa un día.

```python
ventas = np.array([
    [12, 15, 11, 18],
    [8, 10, 14, 9],
    [20, 17, 16, 22]
])

print(ventas)
```

## Paso 3. Consultar la forma y las dimensiones

El atributo `shape` devuelve el tamaño de cada eje del arreglo. En una matriz, el primer valor corresponde al número de filas y el segundo al número de columnas. El atributo `ndim` indica cuántas dimensiones tiene el arreglo, mientras que `size` devuelve el número total de elementos.

```python
print("Forma de la matriz:", ventas.shape)
print("Número de dimensiones:", ventas.ndim)
print("Número total de datos:", ventas.size)
```

## Paso 4. Acceder a una fila

Los índices en Python comienzan en cero. Para obtener una fila completa, se indica su índice entre corchetes. Por ejemplo, `ventas[0]` devuelve la primera fila, que contiene las ventas del Producto A durante los cuatro días.

```python
print("Ventas del Producto A:", ventas[0])
```

## Paso 5. Acceder a un dato por fila y columna

Para seleccionar un solo valor de una matriz se utiliza la forma `arreglo[fila, columna]`. El primer índice identifica la fila y el segundo la columna. Como los índices comienzan en cero, el miércoles corresponde a la columna 2. La instrucción siguiente obtiene las ventas del Producto B el miércoles.

```python
print("Ventas del Producto B el miércoles:", ventas[1, 2])
```

## Paso 6. Seleccionar varias columnas

El *slicing* permite seleccionar una sección de una matriz. La expresión `ventas[:, 1:3]` utiliza `:` para incluir todas las filas y `1:3` para seleccionar las columnas con índices 1 y 2. El límite final no se incluye, por lo que se obtienen las ventas del martes y miércoles para los tres productos.

```python
ventas_martes_miercoles = ventas[:, 1:3]

print(ventas_martes_miercoles)
```

## Paso 7. Calcular las ventas totales de cada producto

La función `np.sum()` suma los valores de un arreglo. Con el argumento `axis=1`, NumPy realiza la suma a lo largo de las columnas de cada fila; en este caso, obtiene el total vendido por cada producto durante los cuatro días. El resultado es un arreglo con un total por producto.

```python
totales_por_producto = np.sum(ventas, axis=1)

print("Total por producto:", totales_por_producto)
```

## Paso 8. Calcular las ventas totales de cada día

Con `axis=0`, la suma se realiza a lo largo de las filas, de modo que se obtiene un total para cada columna o día. Esta operación permite comparar el volumen de ventas diario de la tienda.

```python
totales_por_dia = np.sum(ventas, axis=0)

print("Total por día:", totales_por_dia)
```

## Paso 9. Calcular el promedio de ventas por producto

La función `np.mean()` calcula la media aritmética. Al indicar `axis=1`, se calcula el promedio de cada fila, es decir, el promedio diario de ventas de cada producto durante los cuatro días registrados.

```python
promedios_por_producto = np.mean(ventas, axis=1)

print("Promedio diario por producto:", promedios_por_producto)
```

## Paso 10. Actividad

Con base en los resultados obtenidos, identifica qué producto tuvo el mayor total de ventas y qué día registró la tienda el mayor total. Para comprobar tu interpretación, ejecuta el código siguiente: `np.argmax()` devuelve el índice del valor máximo; recuerda que los índices comienzan en cero.

```python
producto_mayor = np.argmax(totales_por_producto)
dia_mayor = np.argmax(totales_por_dia)

nombres_productos = ["Producto A", "Producto B", "Producto C"]
nombres_dias = ["Lunes", "Martes", "Miércoles", "Jueves"]

print("Producto con mayor venta total:", nombres_productos[producto_mayor])
print("Día con mayor venta total:", nombres_dias[dia_mayor])
```
