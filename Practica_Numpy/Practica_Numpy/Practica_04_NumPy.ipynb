# Práctica 4. Estadísticas por ejes, transformación y combinación de arreglos

Se analizarán resultados de tres indicadores registrados en cuatro periodos. La práctica se centra en estadísticas por eje, reorganización de matrices y combinación de arreglos.

### 1. Importar NumPy

Importa NumPy para trabajar con matrices y funciones estadísticas.

```python
import numpy as np
```

### 2. Crear una matriz de indicadores

Cada fila representa un indicador y cada columna un periodo. Los valores son mediciones ilustrativas.

```python
indicadores = np.array([[72, 75, 78, 81], [64, 69, 71, 74], [80, 78, 83, 86]])
print(indicadores)
```

### 3. Calcular estadísticas globales

Las funciones `mean`, `median`, `std`, `min` y `max` describen el conjunto completo.

```python
print('Promedio:', np.mean(indicadores))
print('Mediana:', np.median(indicadores))
print('Desviación estándar:', np.std(indicadores))
print('Mínimo:', np.min(indicadores))
print('Máximo:', np.max(indicadores))
```

### 4. Calcular promedios por fila

Con `axis=1`, la operación se realiza por fila. Aquí se obtiene el promedio de cada indicador entre periodos.

```python
promedio_indicador = np.mean(indicadores, axis=1)
print(promedio_indicador)
```

### 5. Calcular promedios por columna

Con `axis=0`, se combinan los valores de cada columna entre filas. Aquí se calcula el promedio de los indicadores en cada periodo.

```python
promedio_periodo = np.mean(indicadores, axis=0)
print(promedio_periodo)
```

### 6. Calcular dispersión por indicador

`np.std(..., axis=1)` calcula la desviación estándar de cada fila y permite comparar su variación entre periodos.

```python
dispersion_indicador = np.std(indicadores, axis=1)
print(dispersion_indicador)
```

### 7. Transponer una matriz

La propiedad `.T` intercambia filas y columnas, lo que permite reorganizar la orientación de los datos.

```python
indicadores_transpuestos = indicadores.T
print(indicadores_transpuestos)
print('Dimensiones:', indicadores_transpuestos.shape)
```

### 8. Redimensionar un arreglo

`reshape` cambia la forma sin modificar el número de elementos. El tamaño debe ser compatible con la nueva forma.

```python
secuencia = np.arange(1, 13)
matriz_3x4 = secuencia.reshape(3, 4)
print(matriz_3x4)
```

### 9. Combinar matrices verticalmente

`np.vstack` apila arreglos por filas. Las matrices deben tener el mismo número de columnas.

```python
grupo_a = np.array([[70, 75], [80, 85]])
grupo_b = np.array([[68, 72], [88, 91]])
combinados_vertical = np.vstack((grupo_a, grupo_b))
print(combinados_vertical)
```

### 10. Combinar matrices horizontalmente

`np.hstack` concatena arreglos por columnas. En este ejemplo se combinan mediciones de los mismos registros.

```python
periodo_1 = np.array([[70], [80], [68]])
periodo_2 = np.array([[75], [85], [72]])
combinados_horizontal = np.hstack((periodo_1, periodo_2))
print(combinados_horizontal)
```
