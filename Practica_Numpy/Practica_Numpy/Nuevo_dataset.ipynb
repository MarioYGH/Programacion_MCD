## RÉPLICA DE PRÁCTICA 2: Arreglos Bidimensionales con NumPy

En la Práctica 2 se analizaron ventas estructuradas en matrices (filas = productos, columnas = días).
La hoja **`Resumen`** del nuevo dataset cuenta exactamente con esta forma: **8 productos** en filas y **7 días de la semana** en columnas.

### Pasos 2 y 3 (Nueva Base): Crear matriz 2D y consultar dimensiones

```python
# Extraemos la matriz numérica con las columnas de los 7 días
matriz_ventas = df_resumen.iloc[:, 1:].to_numpy(dtype=int)
nombres_productos = df_resumen['producto'].to_numpy()
nombres_dias = np.array(df_resumen.columns[1:])

print("Matriz de ventas bidimensional (8 productos x 7 días):")
print(matriz_ventas)
print("\nForma de la matriz:", matriz_ventas.shape)
print("Número de dimensiones:", matriz_ventas.ndim)
print("Número total de datos:", matriz_ventas.size)
```

### Paso 4 (Nueva Base): Acceder a una fila completa
Obtener todas las ventas semanales del primer producto (índice 0: Audífonos).

```python
print(f"Ventas semanales de {nombres_productos[0]}:", matriz_ventas[0])
```

### Paso 5 (Nueva Base): Acceder a un dato por fila y columna
Obtener las ventas del producto en la fila 1 (Bolígrafo azul) durante el miércoles (columna 2).

```python
print(f"Ventas de {nombres_productos[1]} el {nombres_dias[2]}:", matriz_ventas[1, 2])
```

### Paso 6 (Nueva Base): Seleccionar varias columnas (Slicing 2D)
Seleccionar las columnas 1 y 2 (Martes y Miércoles) para todos los productos.

```python
ventas_martes_miercoles = matriz_ventas[:, 1:3]
print("Ventas de martes y miércoles para todos los productos:")
print(ventas_martes_miercoles)
```

### Paso 7 (Nueva Base): Ventas totales de cada producto (`axis=1`)
Suma a lo largo de las columnas para colapsar en un total por fila.

```python
totales_por_producto = np.sum(matriz_ventas, axis=1)

print("Ventas totales por producto en la semana:")
for prod, tot in zip(nombres_productos, totales_por_producto):
    print(f" - {prod}: {tot} unidades")
```

### Paso 8 (Nueva Base): Ventas totales de cada día (`axis=0`)
Suma a lo largo de las filas para colapsar en un total por columna.

```python
totales_por_dia = np.sum(matriz_ventas, axis=0)

print("Ventas totales por día de la semana:")
for dia, tot in zip(nombres_dias, totales_por_dia):
    print(f" - {dia}: {tot} unidades")
```

### Paso 9 (Nueva Base): Promedio diario por producto (`axis=1`)

```python
promedios_por_producto = np.mean(matriz_ventas, axis=1)

print("Promedio diario de ventas por producto:")
for prod, prom in zip(nombres_productos, promedios_por_producto):
    print(f" - {prod}: {prom:.2f} unidades/día")
```

### Paso 10 y Ejercicios de Práctica 2 (Nueva Base)
- Identificar producto y día con mayor venta usando `np.argmax()`.
- Ejercicio 1: Ventas del Producto C (fila 2: Botella reutilizable) el martes (columna 1).
- Ejercicio 2: Ventas de lunes y jueves (columnas 0 y 3) para todos los productos.
- Ejercicio 3: Valor máximo de toda la matriz.

```python
# Identificación con argmax
producto_mayor = np.argmax(totales_por_producto)
dia_mayor = np.argmax(totales_por_dia)

print("Producto con mayor venta total:", nombres_productos[producto_mayor], f"({totales_por_producto[producto_mayor]} unidades)")
print("Día con mayor venta total:", nombres_dias[dia_mayor], f"({totales_por_dia[dia_mayor]} unidades)")

# Ejercicio 1: Ventas de Botella reutilizable el martes
print(f"\nVentas de {nombres_productos[2]} el martes:", matriz_ventas[2, 1])

# Ejercicio 2: Ventas de lunes y jueves para todos los productos
ventas_lunes_jueves = matriz_ventas[:, [0, 3]]
print("\nVentas de lunes y jueves:\n", ventas_lunes_jueves)

# Ejercicio 3: Valor máximo de toda la matriz
valor_maximo = np.max(matriz_ventas)
print("\nEl valor máximo de la matriz es:", valor_maximo)
```

## RÉPLICA DE PRÁCTICA 3: Operaciones Vectorizadas, Máscaras Booleanas y Filtrado

En la Práctica 3 se estudió la vectorización, la construcción de máscaras lógicas, el filtrado y el uso de `np.where`.
Replicamos estos métodos utilizando el arreglo de **unidades vendidas** de la hoja `Ventas`.

### Pasos 2 y 3 (Nueva Base): Arreglo y operaciones vectorizadas

```python
unidades = df_ventas['unidades_vendidas'].to_numpy()
print("Arreglo completo de unidades vendidas (56 observaciones):")
print(unidades)

# Operación vectorizada: Proyectar un incremento del 10% sin usar bucles
unidades_ajustadas = unidades * 1.1
print("\nUnidades ajustadas (+10%):")
print(np.round(unidades_ajustadas, 1))
```

### Pasos 4 y 5 (Nueva Base): Máscara booleana y filtrado
Construir una máscara booleana para filtrar ventas superiores a 20 unidades.

```python
mascara = unidades > 20
print("Máscara booleana (unidades > 20):")
print(mascara)

ventas_altas = unidades[mascara]
print("\nObservaciones filtradas (ventas > 20):")
print(ventas_altas)
```

### Paso 6 (Nueva Base): Combinar condiciones lógicas (`&`, `|`)
Filtrar ventas en el rango intermedio: entre 15 y 25 unidades inclusive.

```python
rango_intermedio = unidades[(unidades >= 15) & (unidades <= 25)]
print("Ventas en rango intermedio [15, 25]:")
print(rango_intermedio)
```

### Paso 7 (Nueva Base): Contar observaciones con `np.sum`

```python
cantidad_altas = np.sum(unidades > 20)
print("Cantidad de ventas superiores a 20 unidades:", cantidad_altas)
```

### Pasos 8 y 9 (Nueva Base): Clasificación e índices con `np.where`

```python
# Clasificación condicional
clasificacion = np.where(unidades > 20, "Alta", "Normal/Baja")
print("Clasificación de las primeras 15 ventas:")
print(clasificacion[:15])

# Obtener índices donde se cumple la condición
indices_altos = np.where(unidades > 20)[0]
print("\nÍndices de registros con ventas altas (> 20):")
print(indices_altos)
```

### Paso 10 y Ejercicios de Práctica 3 (Nueva Base)
- Promedio del subconjunto filtrado.
- Ejercicio 1: Filtrar ventas menores que 10.
- Ejercicio 2: Contar cuántas ventas están entre 10 y 20 inclusive.
- Ejercicio 3: Duplicar las ventas (`* 2`).
- Ejercicio 4: Obtener índices de ventas mayores o iguales a 30.

```python
# Promedio de subconjunto filtrado
print("Promedio de ventas altas (> 20):", np.mean(ventas_altas))

# Ejercicio 1: Filtrar ventas menores a 10
mediciones_bajas = unidades[unidades < 10]
print("\nVentas bajas (< 10):", mediciones_bajas)

# Ejercicio 2: Contar ventas en [10, 20]
cantidad_intervalo = np.sum((unidades >= 10) & (unidades <= 20))
print("Cantidad de ventas entre 10 y 20:", cantidad_intervalo)

# Ejercicio 3: Duplicar ventas
unidades_doble = unidades * 2
print("Ventas duplicadas (primeras 8):", unidades_doble[:8])

# Ejercicio 4: Índices con ventas >= 30
indices = np.where(unidades >= 30)[0]
print("Índices de ventas >= 30:", indices)
```

## RÉPLICA DE PRÁCTICA 4: Estadísticas por Ejes, Transformación y Combinación de Arreglos

En la Práctica 4 se abordaron estadísticas globales y por ejes, transposición (`.T`), redimensionamiento (`reshape`) y concatenación (`np.vstack`, `np.hstack`).
Replicamos estos métodos utilizando la matriz bidimensional `matriz_ventas` (8 productos x 7 días).

### Pasos 3 a 6 (Nueva Base): Estadísticas globales y por ejes (`axis=0` y `axis=1`)

```python
print("--- Estadísticas Globales ---")
print("Promedio global:", np.mean(matriz_ventas))
print("Mediana global:", np.median(matriz_ventas))
print("Desviación estándar global:", np.std(matriz_ventas))
print("Mínimo global:", np.min(matriz_ventas))
print("Máximo global:", np.max(matriz_ventas))

print("\n--- Estadísticas por Eje ---")
# Promedio por fila (por producto, axis=1)
promedio_producto = np.mean(matriz_ventas, axis=1)
print("Promedio por producto (axis=1):\n", promedio_producto)

# Promedio por columna (por día, axis=0)
promedio_dia = np.mean(matriz_ventas, axis=0)
print("Promedio por día (axis=0):\n", promedio_dia)

# Desviación estándar por producto (axis=1)
dispersion_producto = np.std(matriz_ventas, axis=1)
print("Dispersión por producto (axis=1):\n", dispersion_producto)
```

### Paso 7 (Nueva Base): Transponer la matriz (`.T`)
Intercambiar filas por columnas para obtener una matriz de días (7) x productos (8).

```python
matriz_transpuesta = matriz_ventas.T
print("Matriz transpuesta (días x productos):")
print(matriz_transpuesta)
print("Dimensiones de la transpuesta:", matriz_transpuesta.shape)
```

### Paso 8 (Nueva Base): Redimensionar un arreglo (`reshape`)

```python
# Secuencia del 1 al 56 (mismo número de elementos que la matriz) redimensionada a 8x7
secuencia = np.arange(1, matriz_ventas.size + 1)
matriz_redimensionada = secuencia.reshape(8, 7)

print("Matriz redimensionada a (8, 7):")
print(matriz_redimensionada)
```

### Pasos 9 y 10 (Nueva Base): Combinar matrices vertical y horizontalmente
- `np.vstack`: apilar arreglos verticalmente (por filas).
- `np.hstack`: concatenar arreglos horizontalmente (por columnas).

```python
# vstack: dividimos la matriz en dos grupos de 4 productos
grupo_a = matriz_ventas[0:4, :]  # Primeros 4 productos
grupo_b = matriz_ventas[4:8, :]  # Últimos 4 productos

combinados_vertical = np.vstack((grupo_a, grupo_b))
print("Forma de vstack:", combinados_vertical.shape)

# hstack: dividimos las columnas entre semana (Lun-Vie) y fin de semana (Sáb-Dom)
dias_laborales = matriz_ventas[:, 0:5] # 5 columnas
fin_de_semana = matriz_ventas[:, 5:7]  # 2 columnas

combinados_horizontal = np.hstack((dias_laborales, fin_de_semana))
print("Forma de hstack:", combinados_horizontal.shape)
```

### Ejercicios de Práctica 4 (Nueva Base)

```python
# Ejercicio 1: Promedio por fila
promedios = np.mean(matriz_ventas, axis=1)
print("Promedios por fila:", promedios)

# Ejercicio 2: Máximo por periodo/día (axis=0)
maximos_periodo = np.max(matriz_ventas, axis=0)
print("Ventas máximas de cada día:", maximos_periodo)

# Ejercicio 3: Arreglo del 1 al 14 redimensionado a matriz de 2x7
datos = np.arange(1, 15)
matriz_2x7 = datos.reshape(2, 7)
print("Matriz 2x7:\n", matriz_2x7)

# Ejercicio 4: Combinar grupo_a y grupo_b verticalmente y mostrar forma
resultado = np.vstack((grupo_a, grupo_b))
print("Forma del resultado vstack:", resultado.shape)
```

## RÉPLICA DE PRÁCTICA 5: Análisis Integrado de Datos Numéricos con NumPy

En la Práctica 5 se integraron validaciones, resúmenes por ejes, máscaras, simulaciones de ajustes y centrado de datos sobre una matriz de 7 días.
Nuestra matriz `matriz_ventas` (8 productos x 7 días) replica con exactitud cada uno de los 12 pasos y los 4 ejercicios.

### Pasos 3 a 5 (Nueva Base): Dimensiones, totales y promedios por producto y por día

```python
print("Dimensiones:", matriz_ventas.shape)
print("Número de datos:", matriz_ventas.size)
print("Número de productos:", matriz_ventas.shape[0])
print("Número de días:", matriz_ventas.shape[1])

# Totales y promedios por producto (fila, axis=1)
total_producto = np.sum(matriz_ventas, axis=1)
promedio_producto = np.mean(matriz_ventas, axis=1)
print("\nTotal semanal por producto:", total_producto)
print("Promedio diario por producto:", np.round(promedio_producto, 2))

# Totales y promedios por día (columna, axis=0)
total_dia = np.sum(matriz_ventas, axis=0)
promedio_dia = np.mean(matriz_ventas, axis=0)
print("\nTotal por día:", total_dia)
print("Promedio por día:", np.round(promedio_dia, 2))
```

### Paso 6 (Nueva Base): Identificar el producto con mayor venta semanal

```python
indice_mayor = np.argmax(total_producto)
print("Índice del producto con mayor venta:", indice_mayor)
print("Producto líder en ventas:", nombres_productos[indice_mayor])
print("Venta semanal máxima:", total_producto[indice_mayor], "unidades")
```

### Pasos 7 y 8 (Nueva Base): Filtrar por umbral y contar días altos por producto

```python
umbral = 20

# Filtrar registros superiores al umbral
ventas_altas_matriz = matriz_ventas[matriz_ventas > umbral]
print(f"Ventas superiores al umbral ({umbral} unidades):")
print(ventas_altas_matriz)
print("Cantidad de registros altos:", ventas_altas_matriz.size)

# Contar cuántos días superó cada producto el umbral (axis=1)
dias_altos_producto = np.sum(matriz_ventas > umbral, axis=1)
print("\nDías con ventas altas (> 20) por cada producto:")
for prod, d in zip(nombres_productos, dias_altos_producto):
    print(f" - {prod}: {d} días")
```

### Paso 9 (Nueva Base): Variabilidad por producto (desviación estándar)

```python
desviacion_producto = np.std(matriz_ventas, axis=1)
print("Desviación estándar por producto:", np.round(desviacion_producto, 2))
```

### Pasos 10 y 11 (Nueva Base): Simular reducción/ajuste y comparar antes y después
Simulamos una reducción de inventario o demanda del 8% (factor 0.92).

```python
ventas_eficientes = matriz_ventas * 0.92
diferencia = matriz_ventas - ventas_eficientes

print("Ventas ajustadas al 92% (primeras 3 filas):")
print(np.round(ventas_eficientes[:3], 1))
print("Diferencia estimada total en unidades:", np.round(np.sum(diferencia), 2))

total_original = np.sum(matriz_ventas)
total_eficiente = np.sum(ventas_eficientes)

print("\nTotal original:", total_original)
print("Total ajustado:", np.round(total_eficiente, 2))
print("Diferencia global:", np.round(total_original - total_eficiente, 2))
```

### Paso 12 (Nueva Base): Centrar los datos por producto
Restar el promedio de cada fila para estudiar las desviaciones respecto al promedio de cada producto.

```python
promedios_fila = np.mean(matriz_ventas, axis=1, keepdims=True)
matriz_centrada = matriz_ventas - promedios_fila

print("Matriz de ventas centrada por producto (primeras 4 filas):")
print(np.round(matriz_centrada[:4], 2))
```

### Ejercicios de Práctica 5 (Nueva Base)
- **Ejercicio 1**: Día de mayor venta conjunta usando `np.argmax(total_dia)`.
- **Ejercicio 2**: Productos con al menos 2 días de venta alta (> 20 unidades) usando `np.where`.
- **Ejercicio 3**: Porcentaje de variación global.
- **Ejercicio 4**: Venta máxima de cada producto y cuál tiene el pico más alto.

```python
# Ejercicio 1: Día de mayor venta total
indice_dia_maximo = np.argmax(total_dia)
print("Índice del día con mayor venta:", indice_dia_maximo)
print("Día con mayor venta conjunta:", nombres_dias[indice_dia_maximo])

# Ejercicio 2: Productos con al menos 2 días de ventas altas
productos_criticos = np.where(dias_altos_producto >= 2)[0]
print("\nÍndices de productos con >= 2 días altos:", productos_criticos)
print("Productos con >= 2 días altos:", nombres_productos[productos_criticos])

# Ejercicio 3: Porcentaje de reducción global
porcentaje_ahorro = ((total_original - total_eficiente) / total_original) * 100
print(f"\nPorcentaje de reducción global: {porcentaje_ahorro:.2f}%")

# Ejercicio 4: Máximo por producto y producto con el pico más alto
maximo_por_producto = np.max(matriz_ventas, axis=1)
print("\nVenta máxima alcanzada por cada producto:", maximo_por_producto)

indice_pico = np.argmax(maximo_por_producto)
print("Producto con la venta máxima más alta:", nombres_productos[indice_pico], f"({maximo_por_producto[indice_pico]} unidades)")
```
